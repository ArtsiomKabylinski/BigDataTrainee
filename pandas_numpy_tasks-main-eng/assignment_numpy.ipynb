**3.** Construct the direct product of arrays (all combinations with each element). A two-dimensional array is provided as input.

```python
def product(arr):
    arr = [np.asarray(a) for a in arr]
    grid = np.meshgrid(*arr, indexing='ij')
    return np.stack(grid, axis=-1).reshape(-1, len(arr))

print(product([[1, 2], [4, 5]]))
```

* __Task 1__: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.  
 For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

```python
def prod_nonzero_py(X):
    result = 1
    for i in range(min(len(X), len(X[0]))):
        if X[i][i] != 0:
            result *= X[i][i]
    return result

def prod_nonzero_np(X):
    d = np.diag(X)
    return d[d != 0].prod()

X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
print(prod_nonzero_py(X.tolist()), prod_nonzero_np(X))
```

* __Task 2__: Two vectors x and y are given. Check if they define the same multiset.  
  For example, for x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) answer is True.

```python
def same_multiset_py(x, y):
    return sorted(x) == sorted(y)

def same_multiset_np(x, y):
    return np.array_equal(np.sort(x), np.sort(y))

x = np.array([1, 2, 2, 4]) 
y = np.array([4, 2, 1, 2])
print(same_multiset_py(x.tolist(), y.tolist()), same_multiset_np(x, y))
```

* __Task 3__: Find the maximum element in the vector x among the elements preceded by zero. 
 For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

```python
def max_after_zero_py(x):
    max_val = 0
    for i in range(1, len(x)):
        if x[i - 1] == 0 and x[i] > max_val:
            max_val = x[i]
    return max_val

def max_after_zero_np(x):
    return x[1:][x[:-1] == 0].max()

x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
print(max_after_zero_py(x.tolist()), max_after_zero_np(x))
```

* __Task 4__: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.  
 For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
def rle_py(x):
    values, counts = [], []
    for v in x:
        if values and values[-1] == v:
            counts[-1] += 1
        else:
            values.append(v)
            counts.append(1)
    return values, counts

def rle_np(x):
    starts = np.r_[0, np.flatnonzero(x[1:] != x[:-1]) + 1] 
    counts = np.diff(np.r_[starts, len(x)])
    return x[starts], counts

x = np.array([2, 2, 2, 3, 3, 3, 5])
print(rle_py(x.tolist()))   
print(rle_np(x)) 
```

* __Task 5__: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

```python
from scipy.spatial.distance import cdist
import timeit

def dist_py(X, Y):
    return [[sum((a - b) ** 2 for a, b in zip(x, y)) ** 0.5 for y in Y] for x in X]

def dist_np(X, Y):
    sq = (X ** 2).sum(1)[:, None] + (Y ** 2).sum(1)[None, :] - 2 * X @ Y.T
    return np.sqrt(np.maximum(sq, 0))

X = np.random.rand(500, 10)
Y = np.random.rand(400, 10)

print(np.allclose(dist_np(X, Y), cdist(X, Y))) 
print(np.allclose(dist_py(X.tolist(), Y.tolist()), cdist(X, Y)))

print("python:", timeit.timeit(lambda: dist_py(X.tolist(), Y.tolist()), number=2) / 2)
print("numpy :", timeit.timeit(lambda: dist_np(X, Y), number=100) / 100)
print("cdist :", timeit.timeit(lambda: cdist(X, Y), number=100) / 100)
```
